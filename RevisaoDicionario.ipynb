'''
6- Complete a função abaixo, que retorne um novo dicionário contendo apenas
os pares cujo valor seja maior ou igual a valor_minimo.
Exemplo: a função filtrar_dicionario({'a': 10, 'b': 15, 'c': 5, 'd': 20}, 10)
deve retornar {'a': 10, 'b': 15, 'd': 20}
'''
def filtrar_dicionario(dicionario, valor_minimo):
  print('filtrar_dicionario(dicionario, valor_minimo)')
  #remova o print acima e escreva o código aqui
#não esqueça de chamar a função (com argumentos, se necessário) aqui abaixo!

```python
def filtrar_dicionario(dicionario, valor_minimo) :
    for chave in list(dicionario.keys()):
        if dicionario[chave] < valor_minimo:
           dicionario.pop(chave)

    return(dicionario)
```

```python
dicionario ={'a': 10, 'b': 15, 'c': 5, 'd': 20}
filtrar_dicionario(dicionario, 15)
```

```python
def filtrar_dicionario2(dicionario, valor_minimo) :
    chaves_a_remover = [chave for chave, valor in dicionario.items() if valor < valor_minimo]

    # Remover as chaves no dicionário
    for chave in chaves_a_remover:
        dicionario.pop(chave)

    return(dicionario)

```

```python
dicionario2 ={'a': 10, 'b': 15, 'c': 5, 'd': 20}

filtrar_dicionario2(dicionario2, 15)


```
